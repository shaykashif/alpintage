```python
%load_ext autoreload
%autoreload 2
import httpx
import pandas as pd
PROD_URL = "https://api.elections.kalshi.com/trade-api/v2"
DEMO_URL = "https://demo-api.kalshi.co/trade-api/v2"
```

```python
query_params = {
    "limit": 200,
    "status": "open",
    "mve_filter": "exclude",   # drop the parlay-style combo markets
    "series_ticker": "KXPLATINUMH"
}
response = (httpx.get(PROD_URL + "/markets", params= query_params, timeout=10))
response.raise_for_status()
data = response.json()
```

```python
data.keys()
len(data["markets"]), data["cursor"]
data["markets"][0]
```

```python
markets = pd.DataFrame(data["markets"])
price_cols = [c for c in markets.columns if c.endswith("_dollars")]
markets[price_cols] = markets[price_cols].apply(pd.to_numeric, errors="coerce")

markets[["ticker", "event_ticker", "yes_bid_dollars", "yes_ask_dollars"]].head(10)
```
